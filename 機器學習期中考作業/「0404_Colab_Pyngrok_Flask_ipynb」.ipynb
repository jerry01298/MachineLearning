```python
from google.colab import userdata

ngrok_authtoken = userdata.get('NGROK_AUTHTOKEN') #從userdata中取得Ngrok的驗證金鑰
port=5051
```

```python
import os
from pyngrok import ngrok #匯入Ngrok工具用來把本機伺服器公開到網路上
```

```python
ngrok.kill()
```

```python
ngrok.set_auth_token(ngrok_authtoken) #設定Ngrok的驗證金鑰
tunnel = ngrok.connect(port, name="linebot_tunnel") #建立一個Ngrok通道把本機的port對外開放
print(tunnel.public_url) #印出對外公開的網址
```

```python
from flask import Flask #匯入Flask框架

app = Flask(__name__) #建立Flask應用程式實例

@app.route('/') #設定首頁路由當使用者進入/網址時會執行這個函式
def home():
    return "<h1>B12090082 楊易桀連線成功</h1>" #回傳一段HTML顯示連線成功訊息

if __name__ == '__main__':
    app.run(port=port)
```
